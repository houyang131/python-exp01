```python
#1 say_hi 无参数，打印 Hello, world!
def say_hi():
    print("Hello, world!")

#2 goat_pad：字符串前后拼接goat并打印
def goat_pad(s):
    print("goat" + s + "goat")

#3 print_n：打印字符串s一共n次，每行一次
def print_n(s, n):
    for _ in range(n):
        print(s)

# --------测试--------
print("====测试say_hi====")
say_hi()

print("\n====测试goat_pad====")
goat_pad("apple")

print("\n====测试print_n====")
print_n("test line", 3)
```

```python
#1 欧几里得算法求最大公约数gcd(m,n)
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

#2 三组测试用例
print("====GCD测试====")
print("gcd(2018,2019):", gcd(2018, 2019))
print("gcd(1200,300):", gcd(1200, 300))
print("gcd(5040,60):", gcd(5040, 60))

#3 测试负数参数
print("\n====负数参数测试====")
print("gcd(-1200, 300):", gcd(-1200, 300))
print("gcd(-20, -8):", gcd(-20, -8))

'''
问题思考：
当输入负数时，取模运算仍然可以计算出公约数，但是结果符号取决于内部运算；
数学上最大公约数应当针对非负整数定义，负数输入得到的结果在数学意义上并不严谨，行为没有实际意义。
'''
```

```python
#1 euler_limit 极限公式 (1+1/n)^n
def euler_limit(n):
    return (1.0 + 1.0 / n) ** n

#2 euler_infinite_sum 级数前n项求和
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    for k in range(n):
        fact = math.factorial(k)
        total += 1.0 / fact
    return total

#3 euler_approx：误差小于ϵ时停止迭代
def euler_approx(epsilon):
    e_true = math.exp(1)
    k = 0
    approx = 0.0
    while abs(approx - e_true) >= epsilon:
        approx += 1.0 / math.factorial(k)
        k += 1
    return approx

#4 print_euler_sum_table：打印1~n每一项的近似值
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val:.6f}")


# --------测试--------
print("====euler_limit测试====")
print(f"n=1000, e近似:{euler_limit(1000):.6f}")

print("\n====euler_infinite_sum测试====")
print(f"n=0 → {euler_infinite_sum(0)}")
print(f"n=1 → {euler_infinite_sum(1)}")
print(f"n=4 → {euler_infinite_sum(4):.6f}")

print("\n====euler_approx 误差小于0.001====")
approx_e = euler_approx(0.001)
print(f"近似e={approx_e:.6f},真实e={math.exp(1):.6f}")

print("\n====print_euler_sum_table====")
print_euler_sum_table(6)
```

```python
#1 判断n是否是2的幂，只用除法和取模，禁止math.sqrt
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

#2 通用：判断 n 是否是 b 的幂
def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

#测试
print("====is_power_of_2测试====")
test_2 = [1,2,4,8,10,16]
for num in test_2:
    print(f"{num} 是2的幂？ {is_power_of_2(num)}")

print("\n====is_power通用测试====")
print("is_power(3, 9):", is_power(3,9))
print("is_power(3,10):", is_power(3,10))
print("is_power(5,125):", is_power(5,125))
print("is_power(2,64):", is_power(2,64))
```
