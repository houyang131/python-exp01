```python
print("====任务1：\\d+ \\s+ \\bcat\\b====")
text1 = "cat on mat; later: cats and dogs"
res_d = re.findall(r"\d+", text1)
res_s = re.findall(r"\s+", text1)
res_cat_word = re.findall(r"\bcat\b", text1)
print("\\d+ 数字：", res_d)
print("\\s+ 空白：", res_s)
print(r"\bcat\b 完整单词cat：", res_cat_word)

print("\n====任务2 分支cat|dog；分支顺序贪婪对比====")
text2 = "cat dog bird CAT"
res_branch1 = re.findall(r"cat|dog", text2)
print("cat|dog：", res_branch1)

s_aaaa = "aaaa"
res_a1 = re.findall(r"a|aa|aaa", s_aaaa)
res_a2 = re.findall(r"aaa|aa|a", s_aaaa)
print("re.findall(r'a|aa|aaa','aaaa') →", res_a1)
print("re.findall(r'aaa|aa|a','aaaa') →", res_a2)
'''
说明：正则分支从左向右匹配，一旦左侧模式匹配成功就不再尝试右侧。
a|aa|aaa优先匹配单字符a；aaa|aa|a优先匹配更长的aaa。
'''

print("\n====任务3 量词 * + ? {m} {m,n}====")
text3 = "ct cat caat caaat"
pats = [
    (r"ca*t", "ca*t：a出现0次或多次"),
    (r"ca+t", "ca+t：a出现1次或多次"),
    (r"ca?t", "ca?t：a出现0次或1次"),
    (r"ca{2}t", r"ca{2}t：a严格2次"),
    (r"ca{1,2}t", r"ca{1,2}t：a1~2次")
]
for pat,desc in pats:
    res = re.findall(pat, text3)
    print(f"{desc} → {res}")

'''
量词含义
*：前字符0次或更多；
+：前字符1次或更多；
?：前字符0次或1次；
{m}：前字符恰好m次；
{m,n}：前字符m~n次（包含m,n）
'''
```

```python
print("====任务1捕获组提取邮箱用户名、域名====")
mail_text = "联系:alice@example.com;备选:bob@test.org"
mail_pattern = r"([a-zA-Z0-9]+)@([a-zA-Z0-9.]+)"
mails = re.findall(mail_pattern, mail_text)
for user,domain in mails:
    print(f"用户名:{user}, 域名:{domain}")

print("\n====任务2 后向引用匹配连续重复单词====")
dup_text = "cat cat, dog cat, bird bird"
dup_pat = r"\b(\w+)\b \b\1\b"
dup_res = re.findall(dup_pat, dup_text)
print("连续重复的单词：", dup_res)

print("\n====任务3：日期格式转换 YYYY‑MM‑DD → MM‑DD‑YYYY，兼容‑ / .分隔符====")
date_text = """2024-05-01
2024/05/02
2024.05.03"""
#捕获年、月、日，分隔符 - / .
date_pat = r"(\d{4})[-\./](\d{2})[-\./](\d{2})"
date_new = re.sub(date_pat, r"\2-\3-\1", date_text)
print("转换前：")
print(date_text)
print("\n转换后：")
print(date_new)

print("\n====任务4：电话号码统一格式 (123) 456‑7890====")
phone_text = "联系我:123-456-7890 或 987.654.3210"
phone_pat = r"(\d{3})[-\.](\d{3})[-\.](\d{4})"
phone_new = re.sub(phone_pat,r"(\1) \2-\3", phone_text)
print("原始：",phone_text)
print("转换：",phone_new)
```

```python
html_doc = """
<html><head><title>实验课程</title></head><body> <h1>数据分析</h1> <p class="course">第六讲:正则表达式</p> <p class="course">第七讲:结构化数据</p> <a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a> </body></html>
"""

#1 构建解析树
soup = BeautifulSoup(html_doc, "html.parser")
title_text = soup.title.text
h1_tag = soup.h1
print("页面标题文本：", title_text)
print("h1标签名：", h1_tag.name, "h1文本：", h1_tag.text)

print("\n====任务2 find_all获取class=course段落，所有链接====")
p_course = soup.find_all("p", class_="course")
print("class='course'的p段落：")
for p in p_course:
    print("  ",p.text)

a_links = soup.find_all("a")
print("\n全部链接：")
for a in a_links:
    link_text = a.get_text()
    href = a.get("href")
    print(f"  文字:{link_text}, href:{href}")

print("\n====任务3 get_text()对比文档、段落、链接的文本====")
full_text = soup.get_text()
p_text_sample = p_course[0].get_text()
a_text_sample = a_links[0].get_text()
print("完整文档get_text():")
print(repr(full_text))
print("\n单个p段落get_text():", repr(p_text_sample))
print("单个a链接get_text():", repr(a_text_sample))
```
