```python
#任务1 score_level成绩等级函数
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0~100范围内")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print("score_level测试：",score_level(92),score_level(75),score_level(40))

#任务2 for、while循环求1~100累加和
print("\n=====1‑100求和=====")
sum_for = 0
for i in range(1,101):
    sum_for +=i
print(f"for循环结果：{sum_for}")

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i +=1
print(f"while循环结果：{sum_while}")

#任务3 迭代阶乘 & 递归阶乘
def fact_iter(n):
    if n <0:
        raise ValueError("n不能为负数")
    res = 1
    for k in range(2,n+1):
        res *=k
    return res

def fact_rec(n):
    if n <0:
        raise ValueError("n不能为负数")
    if n == 0:
        return 1
    return n * fact_rec(n-1)

print("\n====阶乘对比====")
for n in [0,1,5,10]:
    print(f"n={n},迭代阶乘={fact_iter(n)},递归阶乘={fact_rec(n)}")

#任务4 返回元组、元组解包
def summary(values):
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)

res_tuple = summary([12,45,78,33])
mi,ma,av = res_tuple
print(f"\nsummary函数：最小{mi},最大{ma},平均值{av:.2f}")

#任务5 可变参数*args
def sum_all(*args):
    return sum(args)
print("\nsum_all可变参数测试")
print("无参数：",sum_all())
print("单参数9：",sum_all(9))
print("多参数1,2,3,4,5：",sum_all(1,2,3,4,5))
```

```python
import os
import pickle

# 直接把函数写在本单元格，不用student_utils.py文件！
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，应有2个，得到{len(parts)}")
    name, score_str = parts
    name = name.strip()
    if len(name) == 0:
        raise ValueError("姓名为空")
    try:
        score = int(score_str.strip())
    except ValueError:
        raise ValueError("成绩无法转为整数")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩{score}超出0-100范围")
    return {"name": name, "score": score}


#1 创建目录
cwd = os.getcwd()
print("当前工作目录：",cwd)
data_dir = os.path.join(cwd,"exp01_data")
os.makedirs(data_dir,exist_ok=True)
txt_path = os.path.join(data_dir,"students.txt")

#2 写入文本文件
txt_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(txt_path,"w",encoding="utf-8") as f:
    f.write(txt_content)

#逐行解析，捕获异常
valid_records = []
error_records = []
with open(txt_path,"r",encoding="utf-8") as f:
    all_lines = f.readlines()

for line_num, one_line in enumerate(all_lines[1:],start=2):
    try:
        res = parse_student(one_line)
        valid_records.append(res)
    except ValueError as e:
        error_records.append((line_num,str(e)))

print("有效记录：",valid_records)
print("错误记录(行号,原因)：",error_records)

#5 report.txt w模式覆盖，a模式追加
valid_cnt = len(valid_records)
invalid_cnt = len(error_records)
if valid_cnt>0:
    avg = sum(i["score"] for i in valid_records)/valid_cnt
    p_cnt = sum(1 for i in valid_records if i["score"]>=60)
    p_rate = p_cnt / valid_cnt
else:
    avg = 0
    p_rate =0

report_path = os.path.join(data_dir,"report.txt")
# w覆盖
with open(report_path,"w",encoding="utf-8") as f:
    f.write(f"有效记录数:{valid_cnt}\n")
    f.write(f"无效记录数:{invalid_cnt}\n")
    f.write(f"平均分:{avg:.2f}\n")
    f.write(f"及格率:{p_rate:.1%}\n")

# a追加
with open(report_path,"a",encoding="utf-8") as f:
    f.write("=====a模式追加测试行=====\n")

#6 pickle保存读取
pkl_file = os.path.join(data_dir,"students.pkl")
with open(pkl_file,"wb") as f:
    pickle.dump(valid_records,f)

with open(pkl_file,"rb") as f:
    load_data = pickle.load(f)
print("\npickle加载数据与原数据相等：", load_data == valid_records)

#7测试parse_student函数
print("\nparse_student有效测试：",parse_student("Lily,77"))

#8捕获文件不存在异常
try:
    with open("not_exist_file.txt","r",encoding="utf-8") as f:
        pass
except FileNotFoundError as e:
    print("\n捕获文件不存在异常：",e)
```

```python
import copy

#1 Student类
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >= 60
    def __str__(self):
        return f"Student(name={self.name},score={self.score})"

#2 由有效记录创建对象
print("====Student对象====")
for item in valid_records:
    s_obj = Student(item["name"],item["score"])
    print(s_obj, "是否及格",s_obj.is_passed())

#3 GraduateStudent子类继承
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic})"

g_stu = GraduateStudent("Zhang",84,"计算机视觉")
print("\n子类对象：",g_stu,"是否及格",g_stu.is_passed())

#4 直接赋值、浅拷贝copy.copy、深拷贝copy.deepcopy
print("\n====深浅拷贝对比====")

#①直接赋值
ori1 = {"name":"Alice","scores":[85,90]}
ref = ori1
ref["scores"].append(100)
print("直接赋值修改内部列表后 ori1:",ori1)

#②浅拷贝，重置原始对象
ori2 = {"name":"Alice","scores":[85,90]}
shallow_cp = copy.copy(ori2)
shallow_cp["scores"].append(99)
print("浅拷贝修改内部列表后 ori2:",ori2)

#③深拷贝，重置原始对象
ori3 = {"name":"Alice","scores":[85,90]}
deep_cp = copy.deepcopy(ori3)
deep_cp["scores"].append(77)
print("深拷贝修改内部列表后 ori3:",ori3)
```
