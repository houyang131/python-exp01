```python
# 1.回文判断 is_palindrome
def is_palindrome(s: str) -> bool:
    # 忽略空格，全部转为小写
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]

# 2.字母顺序 is_abecedarian
def is_abecedarian(s: str) -> bool:
    clean = s.replace(" ", "").lower()
    for i in range(len(clean)-1):
        if clean[i] > clean[i+1]:
            return False
    return True

#3 去除元音 a e i o u
def remove_vowels(s: str) -> str:
    vowels = {"a","e","i","o","u","A","E","I","O","U"}
    res = []
    for ch in s:
        if ch not in vowels:
            res.append(ch)
    return "".join(res)

# --------测试--------
print("====is_palindrome测试====")
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))

print("\n====is_abecedarian测试====")
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))

print("\n====remove_vowels测试====")
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

```python
#1 list_reverse，输入不是列表抛异常
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]

#2 is_sorted 只遍历一次，判断升序
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

#测试
print("====list_reverse====")
print(list_reverse([1,2,3]))
try:
    list_reverse((1,2,3))
except TypeError as e:
    print("捕获异常：", e)

print("\n====is_sorted====")
print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2,4]))
print(is_sorted((2,4,6)))
```

```python
#1 向量标量乘法
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或者浮点数")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组元素必须是int/float")
    return tuple(x * s for x in t)

#2 向量内积
def vec_inner_product(t1, t2):
    if not (isinstance(t1, tuple) and isinstance(t2, tuple)):
        raise TypeError("输入必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量维度必须相等")
    for e in t1+t2:
        if not isinstance(e,(int,float)):
            raise TypeError("向量元素必须是数字")
    total = 0.0
    for a,b in zip(t1,t2):
        total += a*b
    return total

#3 检查有效矩阵（元组的元组）
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    row_len = None
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if row_len is None:
            row_len = len(row)
        else:
            if len(row) != row_len:
                return False
        for elem in row:
            if not isinstance(elem, (int,float)):
                return False
    return True

#4 矩阵向量乘法 mx_vec_mult
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("矩阵格式不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n_col = len(mx[0]) if len(mx)>0 else 0
    if len(vec) != n_col:
        raise ValueError("矩阵列数与向量长度不一致")
    out = []
    for row in mx:
        s = 0.0
        for mr, vr in zip(row, vec):
            s += mr * vr
        out.append(s)
    return tuple(out)

#测试
print("====vec_scalar_mult====")
print(vec_scalar_mult((1,2,3), 2))

print("\n====vec_inner_product====")
print(vec_inner_product((1,2),(3,4)))

print("\n====check_valid_mx====")
mx_ok = ((1,2),(3,4))
print(check_valid_mx(mx_ok))

print("\n====mx_vec_mult====")
print(mx_vec_mult(mx_ok, (5,6)))
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        # dim：维度；elements列表/元组
        if not isinstance(dim, int):
            raise TypeError("维度必须为整数")
        if dim < 0:
            raise ValueError("维度不能是负数")
        # 实例属性
        self.dim = dim
        if elements is None:
            # 默认全部0
            self.elements = [0.0]*dim
        else:
            if len(elements) != dim:
                raise ValueError("元素数量和维度不一致")
            self.elements = list(elements)

    # 2.说明：dim 和 elements 使用实例属性
    '''
    选择实例属性理由：
    每个Vector实例拥有自己独立的维度与元素；不同向量维度、元素可以不一样。
    如果作为类属性，所有对象共享同一套维度和数据，无法表示多个不同向量。
    '''

    # 3.比较运算符： == < <= > >=
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("维度不同不能比较")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同不能比较")
        return self.elements < other.elements

    def __le__(self, other):
        return self < other or self == other

    def __gt__(self, other):
        return not self <= other

    def __ge__(self, other):
        return not self < other

    #4 dot内积方法
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("向量维度不一致，无法求内积")
        res = 0.0
        for a,b in zip(self.elements, other.elements):
            res += a*b
        return res

    #5 乘法：标量左右乘；向量逐元素相乘
    def __mul__(self, other):
        # v * other
        if isinstance(other, (int, float)):
            new_elems = [x*other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度不一致，不能逐元素相乘")
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("不支持该类型乘法")

    def __rmul__(self, scalar):
        # 2 * v 标量左乘
        return self * scalar

    def __repr__(self):
        return f"Vector(dim={self.dim}, elements={self.elements})"


# --------测试用例--------
print("====Vector类测试====")
v1 = Vector(2, [2,4])
v2 = Vector(2, [2,5])
print(v1)
print(v1 < v2)
print(v1.dot(v2))

v3 = Vector(2,[2,3])
v4 = Vector(2,[4,5])
print("v3*v4逐元素相乘：", v3*v4)
print("标量右乘 v3*2：", v3*2)
print("标量左乘 2*v3：", 2*v3)
```
